# Runtime Comparisons for Constructing Alignments

```python
df12 = pd.read_csv("results/Results_BPI_2012/Comparison_IBF_vs_TRIE_on_playout_with_opt_alignments_BPI_Challenge_2012.csv", delimiter=",")

###Preparation Phase
#"playout_time_ms_adjusted" --> playout time
#"mixed_ibf_creation_time_adjusted" --> ibf creation + binning
df12["mixed_ibf_creation_time_adjusted"] = df12["mixed_ibf_creation_time"] / 1000
df12["playout_time_ms_adjusted"] = df12["playout_time_ms"] / 1000
df12["mixed_ibf_preparation_time"] = (df12["playout_time_ms"]+df12["mixed_ibf_creation_time"]) / 1000

###Conformance Checking Phase
#"mixed_ibf_search_time_adjusted" --> entire alignment construction algorithm
#"mixed_ibf_trace_calculation_time_adjusted" --> IBF access time in algorithm
df12["mixed_ibf_search_time_adjusted"] = df12["mixed_ibf_search_time"] / 1000
df12["mixed_ibf_trace_calculation_time_adjusted"] = df12["mixed_ibf_trace_calculation_time"] / 1000

###Compared against
#"trie_mixed_alignment_time_ms_adjusted" --> only alignment time
df12["trie_mixed_alignment_time_ms_adjusted"] = df12["trie_mixed_alignment_time_ms"] / 1000

df12 = df12[df12["playout_variants"]== 1000]
df12 = df12[df12["noise"]== 0.2]
df12 = df12[df12["buckets_calc"]== "n"]
df12 = df12[df12["omh_w"].isin([5,0])]

df12.groupby(["bucketing","k", "omh_w"])[("mixed_ibf_preparation_time")].agg(["mean", "std"]).reset_index().round(2)#.describe()#["mixed_ibf_search_time"].agg(["mean", "std"])
#df12.groupby(["bucketing","k"])["trie_mixed_alignment_time_ms_adjusted"].agg(["mean", "std"]).reset_index().round(3)#.describe()#["mixed_ibf_search_time"].agg(["mean", "std"])
```
